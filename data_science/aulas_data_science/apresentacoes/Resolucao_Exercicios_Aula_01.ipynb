## 1. Importação de Bibliotecas

Pandas será usado para organizar e analisar as tabelas; NumPy para operações numéricas; Matplotlib para os gráficos.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

plt.style.use("seaborn-v0_8-whitegrid")
```

## 2. Leitura e Visualização dos Dados

Os dados dos exercícios foram fornecidos no enunciado, então serão montados diretamente em DataFrames. Para outros arquivos, Pandas também permite usar `pd.read_csv("arquivo.csv")` ou `pd.read_excel("arquivo.xlsx")`.

A primeira inspeção pode ser feita com `head()`.

```python
temperaturas = {
    "tempo": ["08:00", "09:00", "10:00", "11:00", "12:00", "13:00", "14:00", "15:00", "16:00", "17:00"],
    "temperatura": [25, 28, 32, 35, 38, 42, 45, 41, 37, 33],
}

vendas = {
    "cliente": ["C1", "C2", "C3", "C4", "C5", "C6", "C7", "C8", "C9", "C10"],
    "setor": ["Tecnologia", "Indústria", "Comércio", "Tecnologia", "Indústria", "Comércio", "Tecnologia", "Indústria", "Comércio", "Tecnologia"],
    "vendas": [1500, 2500, 1200, 3000, 4500, 2000, 3500, 5000, 2800, 4000],
}

acoes = {
    "dia": ["D1", "D2", "D3", "D4", "D5", "D6", "D7", "D8", "D9", "D10"],
    "preco": [25.50, 26.20, 25.80, 27.00, 28.50, 27.80, 29.20, 30.50, 29.80, 31.00],
}

df_temperatura = pd.DataFrame(temperaturas)
df_vendas = pd.DataFrame(vendas)
df_acoes = pd.DataFrame(acoes)

df_temperatura.head()
```

## 3. Análise Exploratória dos Dados

Nesta etapa são verificadas as dimensões, os tipos de dados, estatísticas descritivas, médias e medianas. Também são contados valores ausentes e registros repetidos.

```python
for nome, tabela in {
    "Temperatura": df_temperatura,
    "Vendas": df_vendas,
    "Ações": df_acoes,
}.items():
    print(f"\n{nome}: {tabela.shape[0]} linhas e {tabela.shape[1]} colunas")
    print("Tipos de dados:")
    print(tabela.dtypes)
    print("Valores ausentes por coluna:")
    print(tabela.isna().sum())
    print("Linhas duplicadas:", tabela.duplicated().sum())

print("\nEstatísticas de temperatura:")
display(df_temperatura["temperatura"].describe())
print("Média:", df_temperatura["temperatura"].mean())
print("Mediana:", df_temperatura["temperatura"].median())

print("\nVendas somadas e médias por setor:")
display(df_vendas.groupby("setor")["vendas"].agg(["sum", "mean", "median"]))

print("\nEstatísticas dos preços:")
display(df_acoes["preco"].describe())
```

## 4. Limpeza e Preparação dos Dados

Os dados do enunciado não apresentam valores ausentes nem linhas repetidas. A célula abaixo remove duplicatas caso existam e garante que as colunas de análise sejam numéricas, sem descartar observações válidas.

```python
antes = {
    "Temperatura": len(df_temperatura),
    "Vendas": len(df_vendas),
    "Ações": len(df_acoes),
}

df_temperatura = df_temperatura.drop_duplicates().copy()
df_vendas = df_vendas.drop_duplicates().copy()
df_acoes = df_acoes.drop_duplicates().copy()

df_temperatura["temperatura"] = pd.to_numeric(df_temperatura["temperatura"], errors="coerce")
df_vendas["vendas"] = pd.to_numeric(df_vendas["vendas"], errors="coerce")
df_acoes["preco"] = pd.to_numeric(df_acoes["preco"], errors="coerce")

for nome, tabela in {
    "Temperatura": df_temperatura,
    "Vendas": df_vendas,
    "Ações": df_acoes,
}.items():
    removidas = antes[nome] - len(tabela)
    print(f"{nome}: {removidas} duplicata(s) removida(s); {tabela.isna().sum().sum()} valor(es) ausente(s).")
```

## 5. Criação de Visualizações

A primeira figura mostra a variação da temperatura; a segunda compara o faturamento total por setor; a terceira apresenta os preços de fechamento das ações.

```python
faturamento_setor = df_vendas.groupby("setor")["vendas"].sum()

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

df_temperatura.plot(x="tempo", y="temperatura", marker="o", legend=False, ax=axes[0])
axes[0].set_title("Temperatura da máquina")
axes[0].set_xlabel("Horário")
axes[0].set_ylabel("Temperatura (°C)")

faturamento_setor.plot(kind="bar", color="#33766a", ax=axes[1])
axes[1].set_title("Faturamento por setor")
axes[1].set_xlabel("Setor")
axes[1].set_ylabel("Faturamento (R$)")
axes[1].tick_params(axis="x", rotation=20)

df_acoes.plot(x="dia", y="preco", marker="o", legend=False, ax=axes[2])
axes[2].set_title("Preço de fechamento das ações")
axes[2].set_xlabel("Dia")
axes[2].set_ylabel("Preço (R$)")

plt.tight_layout()
plt.show()
```

## 6. Aplicação de Funções Estatísticas

Abaixo são calculadas média, mediana, desvio padrão, variações entre observações e correlação entre o número do dia e o preço. As respostas consideram os valores exibidos pelo enunciado.

```python
df_temperatura["variacao"] = df_temperatura["temperatura"].diff()
maior_elevacao = df_temperatura["variacao"].max()
intervalos_maior_elevacao = df_temperatura.index[df_temperatura["variacao"] == maior_elevacao]

df_acoes["variacao"] = df_acoes["preco"].diff()
quedas = df_acoes.loc[df_acoes["variacao"] < 0, ["dia", "preco", "variacao"]]
df_acoes["dia_numero"] = np.arange(1, len(df_acoes) + 1)

print("EXERCÍCIO 1")
print("Maior temperatura:", df_temperatura["temperatura"].max(), "°C")
print("Menor temperatura:", df_temperatura["temperatura"].min(), "°C")
print("Média:", df_temperatura["temperatura"].mean(), "°C")
print("Mediana:", df_temperatura["temperatura"].median(), "°C")
print("Desvio padrão:", df_temperatura["temperatura"].std())
print("Maior elevação:", maior_elevacao, "°C")
for indice in intervalos_maior_elevacao:
    print("Intervalo:", df_temperatura.loc[indice - 1, "tempo"], "a", df_temperatura.loc[indice, "tempo"])

print("\nEXERCÍCIO 2")
print("Faturamento por setor:\n", df_vendas.groupby("setor")["vendas"].sum())
print("Média por setor:\n", df_vendas.groupby("setor")["vendas"].mean())
print("Faturamento total:", df_vendas["vendas"].sum())
print("Maior venda:\n", df_vendas.loc[df_vendas["vendas"].idxmax()])

print("\nEXERCÍCIO 3")
print("Maior preço: R$", df_acoes["preco"].max())
print("Menor preço: R$", df_acoes["preco"].min())
print("Média: R$", df_acoes["preco"].mean())
print("Mediana: R$", df_acoes["preco"].median())
print("Desvio padrão: R$", df_acoes["preco"].std())
print("Dias com queda:\n", quedas)
print("Correlação dia-preço:", df_acoes["dia_numero"].corr(df_acoes["preco"]))
lucro = (df_acoes["preco"].iloc[-1] - df_acoes["preco"].iloc[0]) * 100
print("Lucro bruto para 100 ações: R$", lucro)
```
